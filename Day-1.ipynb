```python
# calculate Percentage
subject1 = input("enter the name of subject")
marks1 = int(input(f"enter the marks of {subject1}"))

subject2 = input("enter the name of subject")
marks2 = int(input(f"enter the marks of {subject2}"))

subject3 = input("enter the name of subject")
marks3 = int(input(f"enter the marks of {subject3}"))

subject4 = input("enter the name of subject")
marks4 = int(input(f"enter the marks of {subject4}"))

subject5 = input("enter the name of subject")
marks5 = int(input(f"enter the marks of {subject5}"))

subject6 = input("enter the name of subject")
marks6 = int(input(f"enter the marks of {subject6}"))

total = 600
sum = marks1 + marks2 + marks3 + marks4 + marks5 + marks6
percent = sum/total *100

if sum > 90:
    print(f"you have scored {sum}, and your Grade is A+")
elif sum > 80:
     print(f"you have scored {sum}, and your Grade is A")
elif sum > 70:
     print(f"you have scored {sum}, and your Grade is B")
elif sum > 60:
     print(f"you have scored {sum}, and your Grade is C")

elif sum > 50:
     print(f"you have scored {sum}, and your Grade is D")

elif sum < 50:
     print(f"you have scored {sum}, your percentage is {percent}, and your Grade is F")
else:
    print("invalid score")
```
